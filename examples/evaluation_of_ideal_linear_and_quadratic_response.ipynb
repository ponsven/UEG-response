```python
import numpy as np
import matplotlib.pyplot as plt
import UEG_response as ur
```

```python
# Conditions
rs_str    = "3.23"
theta_str = "1.0"
rs = float(rs_str)
theta = float(theta_str)
inv_theta = 1/theta

# Units
hbar = 1.0
aB = 1.0
m = 1.0
e = 1.0
eps0 = 1/(4*np.pi)

# Normalisation
qF = (9*np.pi/4)**(1/3) / (rs*aB)
EF = hbar**2 * qF**2 / (2*m)
beta = 1/(theta*EF)
n = 3/(4*np.pi*rs**3)
omega_p = np.sqrt(n*e**2/(m*eps0))

# Input 
omega = np.linspace(-5.0, 5.0, 100) * EF/hbar
k = 0.5 * qF

# Tolerances
reltol = 1e-12
abstol = 1e-8
eta_log  = 1e-6
eta_sqrt = 1e-6
eta_pol = 1e-6
points_n = 5
tol_upper = 1e-8
ms = 2
dx = 1e-4
lower = 1e-6


linear_chi_0 = ur.ideal_linear_response(omega, k, m, hbar, n, beta, ms=ms,
                          reltol=reltol, abstol=abstol, eta_log=eta_log, tol_upper=tol_upper, points_n=points_n, force_output=True)

# Plot
norm = n*beta
plt.plot(omega*hbar/EF, np.real(linear_chi_0)/norm, '-b', label=r"Real")
plt.plot(omega*hbar/EF, np.imag(linear_chi_0)/norm, '-r', label=r"Imag")

plt.xlabel(r"$\omega$ [$E_F / \hbar$]")
plt.ylabel(r"$\chi_0^{(1)}(\vec{k}, \omega)$ [$n\beta$]")
plt.title(r"$|\vec{k}|/q_F = %g$"%(k/qF))
plt.legend()

```

```python
# Input 
omega1 = np.linspace(-5.0, 5.0, 100) * EF/hbar
omega2 = 0.0
k1 = 0.5 * qF
k2 = 1.0 * qF
csTheta = 0.7

quadratic_chi_0 = ur.ideal_quadratic_response(omega1, k1, omega2, k2, csTheta, m, hbar, n, beta, ms=ms,
                                              reltol=reltol, abstol=abstol, eta_pol=eta_pol, eta_sqrt=eta_sqrt, eta_log=eta_log, tol_upper=tol_upper,
                                              force_output=True) 


# Plot
norm = n*beta**2
plt.plot(omega1*hbar/EF, np.real(quadratic_chi_0)/norm, '-b', label=r"Real")
plt.plot(omega1*hbar/EF, np.imag(quadratic_chi_0)/norm, '-r', label=r"Imag")

plt.xlabel(r"$\omega_1$ [$E_F / \hbar$]")
plt.ylabel(r"$\chi_0^{(2)}(\vec{k}_1, \omega_1, \vec{k}_2, \omega_2)$ [$n\beta^{2}$]")
plt.title(r"$|\vec{k}_1|/q_F = %g$, $|\vec{k}_2|/q_F = %g$, $\hbar\omega_2/E_F = %g$, $\cos\theta = %g$"%(k1/qF, k2/qF, hbar*omega2/EF, csTheta))
plt.legend()
```
